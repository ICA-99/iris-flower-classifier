```python
# ============================================================
# IRIS DATASET - COMPLETE EDA
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# ------------------------------------------------------------
# 2. LOAD DATASET
# ------------------------------------------------------------

df = pd.read_csv("Iris.csv")


# ------------------------------------------------------------
# 3. BASIC INFORMATION
# ------------------------------------------------------------

print("=" * 60)
print("FIRST 5 ROWS")
print("=" * 60)

print(df.head())


print("\n" + "=" * 60)
print("LAST 5 ROWS")
print("=" * 60)

print(df.tail())


print("\n" + "=" * 60)
print("SHAPE")
print("=" * 60)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])


print("\n" + "=" * 60)
print("COLUMN NAMES")
print("=" * 60)

print(df.columns.tolist())


# ------------------------------------------------------------
# 4. DATA TYPES
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("DATA TYPES")
print("=" * 60)

print(df.dtypes)


# ------------------------------------------------------------
# 5. DATASET INFORMATION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("DATASET INFO")
print("=" * 60)

df.info()


# ------------------------------------------------------------
# 6. STATISTICAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("STATISTICAL SUMMARY")
print("=" * 60)

print(df.describe())


# ------------------------------------------------------------
# 7. CHECK MISSING VALUES
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("MISSING VALUES")
print("=" * 60)

print(df.isnull().sum())


# ------------------------------------------------------------
# 8. CHECK DUPLICATES
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("DUPLICATE ROWS")
print("=" * 60)

print("Number of duplicate rows:", df.duplicated().sum())


# ------------------------------------------------------------
# 9. CHECK UNIQUE VALUES
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("UNIQUE VALUES")
print("=" * 60)

for column in df.columns:
    print(f"\n{column}:")
    print(df[column].unique())


# ------------------------------------------------------------
# 10. TARGET DISTRIBUTION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("TARGET DISTRIBUTION")
print("=" * 60)

print(df["Species"].value_counts())


print("\nTarget percentage:")
print(df["Species"].value_counts(normalize=True) * 100)


# ------------------------------------------------------------
# 11. TARGET DISTRIBUTION - BAR CHART
# ------------------------------------------------------------

df["Species"].value_counts().plot(
    kind="bar"
)

plt.title("Number of Samples per Species")
plt.xlabel("Species")
plt.ylabel("Number of Samples")
plt.xticks(rotation=0)
plt.show()


# ------------------------------------------------------------
# 12. HISTOGRAMS
# ------------------------------------------------------------

numeric_columns = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]

df[numeric_columns].hist(
    figsize=(10, 8),
    bins=15
)

plt.suptitle("Distribution of Numerical Features")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 13. BOXPLOTS
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

df[numeric_columns].boxplot()

plt.title("Boxplot of Numerical Features")
plt.ylabel("Value")
plt.xticks(rotation=20)

plt.show()


# ------------------------------------------------------------
# 14. FEATURE MEAN BY SPECIES
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("MEAN OF FEATURES BY SPECIES")
print("=" * 60)

print(
    df.groupby("Species")[numeric_columns].mean()
)


# ------------------------------------------------------------
# 15. FEATURE MEDIAN BY SPECIES
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("MEDIAN OF FEATURES BY SPECIES")
print("=" * 60)

print(
    df.groupby("Species")[numeric_columns].median()
)


# ------------------------------------------------------------
# 16. MINIMUM VALUES BY SPECIES
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("MINIMUM VALUES BY SPECIES")
print("=" * 60)

print(
    df.groupby("Species")[numeric_columns].min()
)


# ------------------------------------------------------------
# 17. MAXIMUM VALUES BY SPECIES
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("MAXIMUM VALUES BY SPECIES")
print("=" * 60)

print(
    df.groupby("Species")[numeric_columns].max()
)


# ------------------------------------------------------------
# 18. CORRELATION MATRIX
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("CORRELATION MATRIX")
print("=" * 60)

correlation = df[numeric_columns].corr()

print(correlation)


# ------------------------------------------------------------
# 19. CORRELATION HEATMAP
# ------------------------------------------------------------

plt.figure(figsize=(8, 6))

plt.imshow(
    correlation,
    interpolation="nearest"
)

plt.colorbar()

plt.xticks(
    range(len(numeric_columns)),
    numeric_columns,
    rotation=45
)

plt.yticks(
    range(len(numeric_columns)),
    numeric_columns
)

plt.title("Feature Correlation")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 20. SCATTER PLOT
# ------------------------------------------------------------

plt.figure(figsize=(8, 6))

for species in df["Species"].unique():

    subset = df[df["Species"] == species]

    plt.scatter(
        subset["PetalLengthCm"],
        subset["PetalWidthCm"],
        label=species
    )

plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Petal Length vs Petal Width")
plt.legend()

plt.show()


# ------------------------------------------------------------
# 21. SEPAL SCATTER PLOT
# ------------------------------------------------------------

plt.figure(figsize=(8, 6))

for species in df["Species"].unique():

    subset = df[df["Species"] == species]

    plt.scatter(
        subset["SepalLengthCm"],
        subset["SepalWidthCm"],
        label=species
    )

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Sepal Length vs Sepal Width")
plt.legend()

plt.show()


# ------------------------------------------------------------
# 22. PAIRWISE RELATIONSHIPS
# ------------------------------------------------------------

from pandas.plotting import scatter_matrix

scatter_matrix(
    df[numeric_columns],
    figsize=(12, 12),
    diagonal="hist"
)

plt.suptitle("Pairwise Feature Relationships")

plt.show()


# ------------------------------------------------------------
# 23. CHECK OUTLIERS USING IQR
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("OUTLIER ANALYSIS")
print("=" * 60)

for column in numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(f"\n{column}")
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))


# ------------------------------------------------------------
# 24. FINAL DATASET SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("FINAL EDA SUMMARY")
print("=" * 60)

print("Dataset shape:", df.shape)

print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nTarget classes:")
print(df["Species"].value_counts())

print("\nNumerical columns:")
print(numeric_columns)

print("\nCorrelation:")
print(df[numeric_columns].corr())
```
